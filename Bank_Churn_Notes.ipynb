```python
# notice 2 columns starting with "Naive_Bayes"  and column "CLIENTNUM" , are not useful for our analysis will drop them
# Kept getting a typo error trying to type out the full Naive_Bayes 
# so looked for a column whose name starts with "Naive_Bayes" and dropped it

naive_bayes_cols = [col for col in df.columns if col.startswith('Naive_Bayes')]

# Drop CLIENTNUM and Naive_Bayes columns
cols_to_drop = ['CLIENTNUM'] + naive_bayes_cols

# checking which columns I'm about to drop for confirmation
print(cols_to_drop)
```

```python
# Drop the columns
df_clean = df.drop(columns=cols_to_drop)
```

```python
# Checking how many columns are left after dropping the unnecessary columns
df_clean.shape
```

```python
# Review data types after cleaning
df_clean.info()
```

```python
# Attriction_flag is the target variablle, but it's currently text: "Existing Customer" and "Attrited Customer". 
# Let's convert it to binary: 0 for Existing Customer and 1 for Attrited Customer
# Checking dtype first so this is safe to re-run without breaking the mapping
if 'Existing Customer' in df_clean['Attrition_Flag'].unique():
    df_clean['Attrition_Flag'] = df_clean['Attrition_Flag'].map({'Existing Customer': 0, 'Attrited Customer': 1})

# checking the conversion
df_clean['Attrition_Flag'].value_counts()
```

```python
# Gender is also text (M/F), let's convert it to binary: 0 for Male and 1 for Female
# Checking dtype first so this is safe to re-run without breaking the mapping
# I found the dtype label isn't reliable here, 
# ( this padnas version shows text columns as "str" instead of "object" , so dtype check silently failed to catch it)
if 'M' in df_clean['Gender'].unique():
    df_clean['Gender'] = df_clean['Gender'].map({'M': 0, 'F': 1})
    
# checking the conversion
df_clean['Gender'].value_counts()
```

```python
# ===== EXPLORATORY DATA ANALYSIS (EDA) =====
# ==== VIZ1: Overall churn distribution (pie chart) =====

# Before comparing  I want to establish the baseline split first, reinforces that churn is a minority class, which is important for later analysis

churn_counts = df_clean['Attrition_Flag'].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(churn_counts, 
        labels=['Existing Customer', 'Attrited Customer'],
        autopct='%1.1f%%', # Will show percentage with 1 decimal place
        colors=['#66b3ff','#ff9999'], 
        startangle=90) # Start the pie chart from the top
plt.title('Overall Customer Churn Distribution')
plt.show()
```

```python
# ==== VIZ 2: Churn rate by Income Category (bar chart) =====

# I'm calculating the churn rate (the percentage of each income group that has churned) for each income category, which is more informative than just the raw counts.
# .mean() on Attribution_FLag works here because it's binary (0 for existing, 1 for churned), so the mean gives the proportion of churned customers in each group.

churn_by_income = df_clean.groupby('Income_Category')['Attrition_Flag'].mean().sort_values(ascending=False)

plt.figure(figsize=(9, 5))
churn_by_income.plot(kind='bar', color='#DD8452')
plt.title('Churn Rate by Income Category')
plt.ylabel('Churn Rate')
plt.xlabel('Income Category')
plt.xticks(rotation=45)
plt.tight_layout()  # Adjust layout to prevent clipping of tick-labels
plt.show()
```

```python
# ===== VIZ 3: Age distribution of churned vs. retained customers (histogram) =====

# Wants to verify whether churned customers are generally younger than retained customers.
# Age is often an important factor in other industries, so I wanted to see if it was a factor here as well.
# Using transparency (alpha) so overlapping bars are still visible.

plt.figure(figsize=(9, 5))

# splitting data by Attrition_Flag before plotting, so we can plot them separately
plt.hist(df_clean[df_clean['Attrition_Flag'] == 0]['Customer_Age'],
         bins=20, alpha=0.6, label='Existing Customer', color='#4c72b0')
plt.hist(df_clean[df_clean['Attrition_Flag'] == 1]['Customer_Age'],
         bins=20, alpha=0.6, label='Attrited Customer', color='#DD8452')

plt.title('Age Distribution: Existing vs. Attrited Customers')
plt.xlabel('Customer Age')
plt.ylabel('Number of Customers')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Finding: the two distributions overlap heavily and both peak in the 40-50 range.
# Unlike what I expected, age on its own doesn't look like a strong driver of churn in this dataset.
# I need to check other features (like transaction behavior) to find what actually separates the two groups.
```

```python
# ==== VIZ 4: Transaction count by churn status (boxplot) =====

# Since age didn't show a strong difference, I want to check if transaction behavior (like number of transactions) is different between churned and retained customers.

plt.figure(figsize=(7, 5))
sns.boxplot(x='Attrition_Flag', y='Total_Trans_Ct', data=df_clean, 
            hue='Attrition_Flag', palette=['#4c72b0', '#DD8452'], legend=False)
plt.title('Total Transaction Count by Churn Status')
plt.xlabel('Attrition Flag (0=Existing, 1=Attrited)')
plt.ylabel('Total Transaction Count')
plt.tight_layout() 
plt.show()
```

```python
# Finding: this is a much clearer signal than age.
# Existing customers have a median around 70 transactions, while attrited customers sit closer to 40 transactions. Which is a real gap between the two groupss.
# this suggests transactions activity is a meaningful predictor of churn, and good candidate for feature engineering in a predictive model.
```

```python
# ==== VIZ 5: Total spend vs transaction count, colored by churn (scatter) =====

# Want to check if churned customers form a visibly distinct cluster when looking at spending amount and transaction count together,
# rather than one variable at a time. This can help identify patterns that might not be obvious when looking at single variables.

plt.figure(figsize=(8, 6))
plt.scatter(df_clean[df_clean['Attrition_Flag'] == 0]['Total_Trans_Ct'],
            df_clean[df_clean['Attrition_Flag'] == 0]['Total_Trans_Amt'],
            alpha=0.4, label='Existing Customer', color='#4c72b0', s=15)
plt.scatter(df_clean[df_clean['Attrition_Flag'] == 1]['Total_Trans_Ct'],
            df_clean[df_clean['Attrition_Flag'] == 1]['Total_Trans_Amt'],
            alpha=0.4, label='Attrited Customer', color='#DD8452', s=15)

plt.title('Transaction Amount vs Transaction Count by Churn Status')
plt.xlabel('Total Transaction Count')
plt.ylabel('Total Transaction Amount ($)')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Finding: attrited customers cluster almost entirely in the low transaction count/ low spend corner.
# There's also a clear High-value segment of existing customers (100+ transactions and $12,500+ spend) that is not at risk of churn.
# This lines up with the earlier box plot in VIZ4: transaction activity is a strong predictor of churn, and this scatter plot reinforces
# that finding by showing how churned customers are concentrated in the low activity segment.
```

```python
# ===== VIZ 6: Correlation heatmap of numeric features =====

# After finding that transaction behavior is a strong predictor of churn, I want to see how other numeric features relate to each other and mostly with Attrition_Flag.
# Selecting only numeric columns for correlation analysis (like Education_Level and Marital_Status ... are text columns) can't be correlated this way since they are not numeric yet.

numeric_cols = df_clean.select_dtypes(include=['int64', 'float64'])

corr_matrix = numeric_cols.corr()

plt.figure(figsize=(14, 10))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap='coolwarm', center=0)
plt.title('Correlation Heatmap of Numeric Features')
plt.tight_layout()
plt.show()
```

```python
## STATISTICAL ANALYSIS

# I have summary stats from .describe() earlier, but that doesn't include mode, and the brief specifically asks for it 
# So I'm calculating central tendency and spread EXPLICITY  for the features I care about most from my EDA findings

key_features = ['Customer_Age', 'Total_Trans_Ct', 'Total_Trans_Amt', 'Credit_Limit']

for col in key_features:
    print(f"--- {col} ---")
    # Mean: the average value,  add up every value and divide by how many there are
    print(f"Mean: {df_clean[col].mean():.2f}")
    # # Median: the middle value when everything is sorted - less affected, by extreme outliers than the mean, so comparing mean vs median, tells me if a column is skewed
    print(f"Median: {df_clean[col].median():.2f}")
    # Mode: the single most frequently occurring value in the column
    print(f"Mode: {df_clean[col].mode()[0]}")
    # Standard deviation: how spread out the values are from the mean, a small number means most values sit close to the mean, a large number means values are spread out widely
    print(f"Std Dev: {df_clean[col].std():.2f}")
    print()
```

```python
## PREPARE DATA FOR MACHINE LEARNING

# === Encoding remaining categorical columns ===
# Income_Category has a natural order (low to high), so I'm mapping
# it manually to preserve that order - a model should know that
# "$120K+" is genuinely higher than "Less than $40K", not just
# a different unrelated category

income_order = {
    'Less than $40K': 0,
    '$40K - $60K': 1,
    '$60K - $80K': 2,
    '$80K - $120K': 3,
    '$120K +': 4,
    'Unknown': -1 # keeping it unknown as its own distinct value
}

df_clean['Income_Category_Encoded'] = df_clean['Income_Category'].map(income_order)

# I'm doing the same for Education_level  because it also has a rough order (uneducated -> post-graduate)
education_order = {
    'Uneducated': 0,
    'High School': 1,
    'College': 2,
    'Graduate': 3,
    'Post-Graduate': 4,
    'Doctorate': 5,
    'Unknown': -1 # keeping it unknown as its own distinct value
}

df_clean['Education_Level_Encoded'] = df_clean['Education_Level'].map(education_order)


# Marial_Status and Card_category don't have a natural order, so I'm using one-hot encoding for them. 
# This creates a new binary column for each category, which is suitable for machine learning models that don't assume any order in categorical variables.
df_clean = pd.get_dummies(df_clean, columns=['Marital_Status', 'Card_Category'], drop_first=True)

# Checking the new columns after encoding
df_clean.columns
```

```python
# ==== Handle Multicollinearity ====

# Credit_Limit and Avg_Open_To_Buy are almost perfectly correlated
# (Avg_Open_To_Buy = Credit_Limit minus spent amount), so I'm droping one
df_model = df_clean.drop(columns=['Avg_Open_To_Buy']) 
```

```python
# ===== Drop original text columns =====

# Encoded versions of these already exist, so the text versions aren't needed anymore. Dropping them to avoid confusion and reduce dimensionality.
df_model = df_model.drop(columns=['Income_Category', 'Education_Level'])    

df_model.info()  # Final check to see the structure of the dataset ready for modeling
```

```python
# ==== Feature selection and train/test split ====

# Attrition_Flag is the target I want to predict. Everything else is a potential feature.
X = df_model.drop(columns=['Attrition_Flag'])
Y = df_model['Attrition_Flag']

# Now splitting into training and testing sets, so I can check how well the model performs on data it hasn't seen before.
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42, stratify=Y
)

# test_size=0.2 means 20% of the data will be used for testing.
# random_state=42 ensures that the split is reproducible.
# stratify=Y keeps the same churn ratio (-16%) in both the train and test sets, which is important for model evaluation.

print("Training set shape:", X_train.shape)
print("Testing set shape:", X_test.shape)
```

```python
## BUILD AND TRAIN THE DECISION TREE MODEL

from sklearn.tree import DecisionTreeClassifier


# Will use max_depth to limit how the tree can grow, which helps prevent overfitting. A very deep tree can memorize the training data but perform poorly on new data.
# random_state=42 ensures that the model's behavior is reproducible

model = DecisionTreeClassifier(max_depth=5, random_state=42)
# This fit the model on the training data only
# The test sets stays untouched until evaluation, so I get an honest measure of performance
model.fit(X_train, Y_train)

print("Model trained.")
```

```python
# Evaluation of the Model

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report

# Prediction on the test set (data the model hasn't seen before)
Y_pred = model.predict(X_test)

# Evaluation metrics
print("Accuracy:", accuracy_score(Y_test, Y_pred))
print("Precision:", precision_score(Y_test, Y_pred))
print("Recall:", recall_score(Y_test, Y_pred))
print("F1 Score:", f1_score(Y_test, Y_pred))

print()
print("Confusion Matrix:")
print(confusion_matrix(Y_test, Y_pred))

print()
print("Classification Report:")
print(classification_report(Y_test, Y_pred))
```

```python
# Feature importance

# A decision tree can tell me which features it actually relied on most to make its splits.
# This lets me check whether the model agrees whith what I found in my EDA.

importances = pd.Series(model.feature_importances_, index=X.columns)
importances = importances.sort_values(ascending=False)

print(importances.head(10))  # Show the top 10 most important features
```

```python
### Interpretation of Feature Importance

# The Decision Tree relies most heavily on `Total_Trans_Ct` (39%) and `Total_Trans_Amt` (26%), together these two features account for over 65% of 
# the model's decision-making.
# This directly confirms the patterns seen in the box plot (VIZ4), scatter plot (VIZ5), and the correlation heatmap (VIZ6) during EDA.
# `Customer_Age`, which I initially expected to be a factor, contributes less than 1%, consistent with the near-zero correlation and overlapping distributions found earlier.
# This agreement between EDA and model behavior is a good sign that the findings are genuine patterns in the dat, not artifacts of the model.
```

```python
# CONCLUSIONS

# ### Key Findings

# - **Churn rate**: 16.1% of customers in this dataset churned (1,627 out 
#   of 10,127), confirming this is an imbalanced classification problem -
#   something I accounted for throughout the analysis and model evaluation.
 
#  - **Transaction activity is the strongest churn signal.** Both the box
#   plot and scatter plot showed a clear separation: existing customers
#   had a median of ~70 transactions vs. ~40 for attrited customers.
#   This was confirmed statistically by the correlation heatmap
#   (-0.37 with churn) and again by the model itself, where
#   `Total_Trans_Ct` accounted for 39% of the Decision Tree's decisions -
#   by far the single most important feature.

# - **Revolving balance and relationship depth matter too.** Customers
#   with lower `Total_Revolving_Bal` and fewer products/accounts
#   (`Total_Relationship_Count`) were more likely to churn, suggesting
#   disengaged, low-balance customers are the highest-risk group.

# - **Age is not a meaningful churn factor in this dataset** - contrary
#   to what I initially expected. The histogram showed heavily overlapping
#   age distributions, the correlation with churn was near zero (0.02),
#   and the model itself gave age less than 1% importance.

# - **Income shows a mild U-shape relationship with churn** - both the
#   lowest income bracket and highest income bracket churned at a
#   slightly higher rate (~17.5%) than middle-income brackets, with
#   customers who didn't disclose income ("Unknown") behaving similarly
#   to the lowest earners.

# ### Model Performance

# A Decision Tree classifier (max depth 5) was trained to predict churn,
# achieving:
# - 93.1% accuracy (vs. an 84% baseline from always predicting "no churn")
# - 82.5% precision, 72.6% recall, 77.3% F1 score

# The model correctly identifies about 3 out of 4 customers who will
# churn, but still misses roughly 1 in 4 (89 false negatives out of 325
# actual churners in the test set). In a real deployment, this is the
# practical cost of the model - customers who would have been flagged
# for retention outreach but weren't.

# ### Recommendations

# - The bank could use transaction activity and revolving balance as
#   early warning indicators, flagging customers whose activity drops
#   for proactive retention outreach.
# - Since recall is the weaker metric here, a bank prioritizing not
#   missing at-risk customers might accept more false positives
#   (unnecessary outreach) in exchange for catching more true churners -
#   this could be tuned by adjusting the model's decision threshold.

# ### Limitations

# - The dataset only captures a snapshot of account activity, not the
#   reasons behind it - it can't explain *why* transaction activity
#   dropped before someone churned.
# - ~10-15% of records had "Unknown" values for Education_Level,
#   Marital_Status, and Income_Category, which were kept as their own
#   category rather than imputed - this preserves data volume but means
#   the model has less certain information for those customers.
# - The model was only tuned to a max depth of 5 for interpretability;
#   a deeper tree or a different algorithm might achieve higher recall
#   at the cost of being harder to explain.

# ### Future Improvements

# - Compare the Decision Tree against other model types (e.g. Logistic
#   Regression, Random Forest) to see if performance - particularly
#   recall - can be improved.
# - Test different `max_depth` values or use cross-validation to tune
#   the model more rigorously rather than a single fixed depth.
# - Investigate the `Total_Relationship_Count` feature further, since it
#   ranked as the third most important - understanding which specific
#   products/services correlate with retention could inform more
#   targeted recommendations.
```
